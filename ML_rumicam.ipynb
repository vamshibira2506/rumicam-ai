```python
import os

path = "/content/drive/MyDrive/Cow_Buffalo_Human"

folders = {
    "cow": "cow",
    "buffalo": " buffalo",
    "human": " human"
}

for class_name, folder_name in folders.items():
    folder_path = os.path.join(path, folder_name)
    print(class_name, "exists:", os.path.exists(folder_path))
```

```python
import os

for class_name, folder_name in folders.items():
    folder_path = os.path.join(path, folder_name)

    count = 0

    for root, dirs, files in os.walk(folder_path):
        for file in files:
            if file.lower().endswith((".jpg", ".jpeg", ".png", ".webp")):
                count += 1

    print(class_name, ":", count, "images")
```

```python
import tensorflow as tf

IMG_SIZE = (224, 224)
BATCH_SIZE = 32

data = tf.keras.utils.image_dataset_from_directory(
    path,
    validation_split=0.2,
    subset="training",
    seed=123,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_names=["cow", " buffalo", " human"]
)

print("Classes:", data.class_names)
```

```python
validation_data = tf.keras.utils.image_dataset_from_directory(
    path,
    validation_split=0.2,
    subset="validation",
    seed=123,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_names=["cow", " buffalo", " human"]
)

print("Validation dataset ready!")
```

```python
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras import layers, models

# Load the pre-trained MobileNetV2 model
base_model = MobileNetV2(
    input_shape=(224, 224, 3),
    include_top=False,
    weights="imagenet"
)

# Freeze the pre-trained layers
base_model.trainable = False

# Create our model
model = models.Sequential([
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.Dense(128, activation="relu"),
    layers.Dropout(0.3),
    layers.Dense(3, activation="softmax")
])

# Compile the model
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()
```

```python
history = model.fit(
    data,
    validation_data=validation_data,
    epochs=10
)
```

```python
model.save("/content/drive/MyDrive/cow_buffalo_human_model.keras")

print("Model saved successfully!")
```

```python
import numpy as np
from tensorflow.keras.utils import load_img, img_to_array

# Get uploaded image name
image_name = list(uploaded.keys())[0]

# Load and resize image
img = load_img(image_name, target_size=(224, 224))

# Convert image to array
img_array = img_to_array(img)

# Add batch dimension
img_array = np.expand_dims(img_array, axis=0)

# Predict
prediction = model.predict(img_array)

# Class names
class_names = ["cow", "buffalo", "human"]

# Get result
index = np.argmax(prediction)
confidence = prediction[0][index] * 100

print("Prediction:", class_names[index])
print("Confidence:", round(confidence, 2), "%")
```
